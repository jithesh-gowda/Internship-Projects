```python
import pandas as pd
```

```python
members = pd.read_csv('members.csv')
purchase = pd.read_csv('purchase.csv')
```

```python
members
```

```python
purchase
```

```python
df = purchase.merge(members,on='customer_id', how='inner')
```

```python
df
```

```python
gold_member = df[df['membership_level'] == 'Gold'] 
gold_member
```

```python
total_spent = gold_member.groupby('customer_id')['amount'].sum().reset_index()
total_spent
```

```python
final_df = total_spent.merge(members,on = 'customer_id')
final_df
```

```python
final_df = final_df[final_df['amount']>250]
final_df
```
